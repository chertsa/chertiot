# Plot your last 24 hours
Your ThingsBoard session is already wired in — this notebook can only see **your** tenant.

```python
import chertiot
chertiot.devices()
```

```python
df = chertiot.telemetry("my-first-device", keys=["temperature", "humidity"], hours=24)
df.describe()
```

```python
ax = df.plot(figsize=(10, 4), title="My device — last 24 h")
ax.set_ylabel("value")
```
