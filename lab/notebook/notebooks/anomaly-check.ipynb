# Simple anomaly check
Flag readings more than 3 standard deviations from the rolling mean.

```python
import chertiot
df = chertiot.telemetry("my-first-device", keys=["temperature"], hours=24)
m = df["temperature"].rolling(20, min_periods=5).mean()
s = df["temperature"].rolling(20, min_periods=5).std()
anomalies = df[(df["temperature"] - m).abs() > 3 * s]
print(f"{len(anomalies)} anomalous points")
ax = df["temperature"].plot(figsize=(10, 4))
anomalies["temperature"].plot(ax=ax, style="ro")
```
